# Active learning

When the system can be simulated, HyDyReg asks it for training trajectories and then checks the learned model with *equivalence queries*: trajectories of the system and of the model from random initial conditions are compared, and the ones that differ (counterexamples) are used to improve the model, until a query finds none.

```python
import matplotlib.pyplot as plt
import hydyreg as hd
from hydyreg.benchmarks import systems, rows
```

```python
system = systems.isolette()
design = hd.SamplingDesign(time_step=0.25, time_range=25, x0s=[[105.5, 80], [97.5, 100]])
config = hd.LearnConfig(method="incremental", seg_tol=0.01,
                        equivalence=hd.EquivalenceConfig(bounds=[[95, 105], [70, 110]], tol=0.05))
result = hd.learn_from_system(system, design, config)
print(result.summary())
```

```python
result.plot("comparison");
```

## Regions of the form "if ... else if ... else"

The benchmark `affine_3modes` has a T-junction of boundaries: *if $x_1 \ge 1$: mode 3; else if $x_2 \ge 2$: mode 1; else: mode 2*. The default (multinomial) classifier can only approximate it; the decision-list classifier represents it exactly. The rows of the thesis experiments are available from `hydyreg.benchmarks.rows`.

```python
row = rows.get_row("incr/affine_3modes")
row.config.classifier.kind = "decision_list"
res_dl = rows.run_row(row)
print(res_dl.summary())
print(res_dl.equations())
```

```python
res_dl.plot("regions", true_labelfun=row.system().labelfun);
```
