# Learning from CSV files

Recorded trajectories (here: two runs of the isolette benchmark, a heater with a thermostat) are read from CSV files, segmented, and learned with both methods of the thesis. The same can be done from the command line:

```
hydyreg fit data/isolette_*.csv --seg-tol 0.01 --out result/ --latex --plots
```

````python
import matplotlib.pyplot as plt
import hydyreg as hd

trajs = hd.Trajectories.from_csv("data/isolette_*.csv")     # header line: t,T_box,T_heater
print(trajs.num_trajectories, "trajectories; variables:", trajs.names, "; time step:", trajs.check_uniform_step())
````

## Segmentation

Change points are where the derivative jumps (a kink of the trajectory). The samples around them are removed; each remaining segment is assumed to lie in one mode.

````python
import copy
seg = copy.deepcopy(trajs)
segments = seg.segment_trajectories(0.01)
hd.plotting.plot_segmentation(seg, segments, names=trajs.names);
````

## Incremental method

````python
result = hd.learn_from_data(trajs, hd.LearnConfig(method="incremental", seg_tol=0.01))
print(result.summary())
print(result.equations())
result.plot("regions");
````

## Optimization-based method (number of modes given)

````python
config = hd.LearnConfig(method="optimization", n_modes=2, seg_tol=0.01,
                        optimization=hd.OptimizationConfig(n_starts=5))
result_opti = hd.learn_from_data(trajs, config)
print(result_opti.equations())
````
